```python
#Converting M,K and other string into numerical value
def size_process(item):
    if str(item)[-1] == "M":
        res = float(str(item).replace("M",""))
        res = 1024*res
        return res
    elif str(item)[-1] == "k":
        res = float(str(item).replace("k",""))
        return res
    else:
        return str(np.nan)
```

```python
#df_copy['Size'] = df_copy['Size'].apply(size_process)
def check_version(version):

    # Return NaN if the value is missing
    if pd.isna(version):
        return np.nan

    # Remove leading and trailing spaces
    version = version.strip()

    # Treat "Varies with device" as a missing value
    if version == "Varies with device":
        return np.nan

    # Remove 'W' from versions like "4.4W"
    version = version.replace("W", "")

    # If the version is given as a range, compare both versions
    if "-" in version:

        # Split the range into two versions and remove extra spaces
        left, right = [v.strip() for v in version.split("-")]

        # Convert version strings into tuples for accurate comparison
        # Example: "2.3.1" -> (2, 3, 1)
        left_tuple = tuple(map(int, left.split(".")))
        right_tuple = tuple(map(int, right.split(".")))

        # Keep the higher Android version
        if left_tuple > right_tuple:
            version = left
        else:
            version = right

        # The above if-else can also be written as:
        # version = left if left_tuple > right_tuple else right

    # Split the selected version into its components
    parts = version.split(".")

    # Convert versions like "7.1.1" to 7.1
    if len(parts) >= 2:
        return float(f"{parts[0]}.{parts[1]}")

    # Convert versions like "8" to 8.0
    else:
        return float(f"{parts[0]}.0")
```
