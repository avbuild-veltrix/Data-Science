# Creating Your Own Cutoff

```python
# By default it is 0.5 but if you wat to decide your own cutoff, you have to use 
#                  Precision,recall, accuracy tradeoff
#                                   OR
#              Sensitivity, Specificity, accuracy tradeoff          
```

```python
from sklearn.metrics import precision_score, recall_score, accuracy_score
```

```python
# Calculate precision recall and accuracy for differentt threshold probabilities
thresholds = np.linspace(0,1,100)
precisions = []
recalls = []
accuracies = []

for threshold in thresholds:
    y_pred_threshold = (y_pred_proba >= threshold).astype(int)
    precision = precision_score(y_test, y_pred_threshold)
    precisions.append(precision)
    recall = recall_score(y_test, y_pred_threshold)
    recalls.append(recall)
    accuracy = accuracy_score(y_test, y_pred_threshold)
    accuracies.append(accuracy)
```

```python
precisions
```

```python
recalls
```

```python
accuracies
```

# Plot

```python
# Precision, recall, and accuracy against threshold probabilities

plt.figure(figsize = (8,6))
plt.plot(thresholds, precisions, label = 'Precision')
plt.plot(thresholds, recalls, label = 'Recall')
plt.plot(thresholds, accuracies, label = 'Accuracy')
plt.xlabel("Threshold Probabilities")
plt.ylabel("Score")
plt.title("Precision,Recall,Accuracy and Threshold Probability")
plt.legend()
plt.grid(True)
plt.show()
```
