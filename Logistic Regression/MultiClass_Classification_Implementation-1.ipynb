```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import make_classification
X,y = make_classification(n_samples = 1000, n_features = 10, n_redundant = 5, n_informative = 5, n_classes = 3, random_state = 1)
```

# OVR

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.30, random_state = 1)
```

```python
from sklearn.linear_model import LogisticRegression
model_ovr = LogisticRegression(multi_class = 'ovr', solver= 'lbfgs')
model_ovr
```

```python
model_ovr.fit(X_train, y_train)
```

```python
y_pred = model_ovr.predict(X_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

# Multinomial

```python
model_multinomial = LogisticRegression(multi_class = 'multinomial', solver = 'lbfgs')
```

```python
model_multinomial.fit(X_train, y_train)
```

```python
y_pred_multinomial = model_multinomial.predict(X_test)
```

```python
print(confusion_matrix(y_test, y_pred_multinomial))
print(accuracy_score(y_test, y_pred_multinomial))
print(classification_report(y_test, y_pred_multinomial))
```
