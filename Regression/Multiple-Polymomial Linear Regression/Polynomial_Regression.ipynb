```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
np.random.seed(1)
```

```python
# Generate few synthetic data points
X = 2*np.random.rand(100, 1)
```

```python
X
```

```python
# Generate few synthetic data points y using X.
y = 4 + 3 * X + 1.5 * X ** 2 + np.random.rand(100, 1)
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25 ,random_state = 1)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly_features = PolynomialFeatures(degree = 2, include_bias = False)
```

```python
X_poly_train = poly_features.fit_transform(X_train)
X_poly_train
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model
```

```python
model.fit(X_poly_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
# Making prediction on train data
X_train_pred = model.predict(X_poly_train)
```

```python
X_train_pred
```

```python
# Performance metrices
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
print("Mean Squared Error", mean_squared_error(y_train, X_train_pred))
```

```python
 plt.scatter(X_train, y_train, label = "Training Data")
```

```python
degree = 2
plt.scatter(X_train, y_train, label = "training data")
X_range = np.linspace(0, 2, 100).reshape(-1, 1)
X_range_poly = poly_features.transform(X_range)
plt.plot(X_range, model.predict(X_range_poly), color='red', label=f'Polynomial Regression (Degree {degree})')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Polynomial Regression')
plt.legend()
plt.show()
```
