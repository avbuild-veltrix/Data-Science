```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = fetch_california_housing()
```

```python
df = pd.DataFrame(data.data, columns = data.feature_names)
```

```python
# Problem Statement >> To predict the median house value for districts
df['Price'] = data.target
```

```python
# Separate X and y -
X = df.iloc[:, :-1] # Independent Variable
X
```

```python
# y = df.iloc[:, -1] >> this will give us the target column in series form
y = df.iloc[:, [-1]]
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.25, random_state = 1)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
#Scaling
# Y should not be scaled
# transform on test data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
# Model training
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model
```

```python
model.fit(X_train, y_train)
```

```python
coefficient = model.coef_
coefficient
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
# Performance metrices
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
print("Mean absolute error", mean_absolute_error(y_pred, y_test))
```

```python
print("Mean squared error", mean_squared_error(y_pred, y_test))
```

```python
print("R Square", r2_score(y_pred, y_test))
```

```python
r_sq = r2_score(y_pred, y_test)
```

```python
# Adjusted R square
Adjusted_R_Square = 1 - (1 - r_sq) * (len(y_test) - 1) / (len(y_test) - X_test.shape[1] - 1)
Adjusted_R_Square
```
