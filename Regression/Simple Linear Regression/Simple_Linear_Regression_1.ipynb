```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Reading the data
#Here we are loading the from scikit learn
from sklearn.datasets import load_diabetes
```

```python
diabetes = load_diabetes()
```

```python
#Problem statement >> To predict disease progression one year after baseline provided 10 feature
# In one year how thw sugar level has changes
print(diabetes['DESCR'])
```

```python
data = pd.DataFrame(diabetes.data, columns = diabetes.feature_names)
data
```

```python
data.head()
```

```python
data.tail()
```

```python
data['target'] = diabetes.target
```

```python
data
```

```python
X = data[["bmi"]]
y = data[["target"]]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, random_state = 1)
```

```python
#Model training
# How to train the model
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
X_train
```

```python
y_train
```

```python
# Training of model
# At the backend it will use gradient descent to give an optimal coefficient
model.fit(X_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
# Visualize the prediction
plt.scatter(X_test, y_test, color = 'black', label = "ACtual Data")
```

```python
# Visualize the prediction
plt.scatter(X_test, y_test, color = 'black', label = "ACtual Data")
plt.plot(X_test, y_pred, color = 'blue', linewidth = 3, label = "Predited Linear Regression")
plt.xlabel("BMI")
plt.ylabel("One year progression target")
plt.title("Linear Regression on diabetes data")
plt.legend()
plt.show()
```

```python
# Model Evaluation
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, y_pred)
```

```python
print("MSE = ", mse)
```
